```python
df.hist(figsize=(12, 8))

plt.suptitle("Histogram of Dataset")
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")
plt.show()
```

```python
numeric_df.plot(
    kind='box',
    subplots=True,
    layout=(3, 5),
    figsize=(15, 10)
)

plt.tight_layout()
plt.show()
```
